### 03. Data Preprocessing & Encoding

**Objective:** Convert categorical variables to numerical format for machine learning model training.

The following columns will be encoded:
- **sex**: Female → 0, Male → 1
- **smoker**: No → 0, Yes → 1
- **day**: Thur → 0, Fri → 1, Sat → 2, Sun → 3
- **time**: Lunch → 0, Dinner → 1

```python
data["sex"] = data["sex"].map({"Female": 0, "Male": 1})
data["smoker"] = data["smoker"].map({"No": 0, "Yes": 1})
data["day"] = data["day"].map({"Thur": 0, "Fri": 1, "Sat": 2, "Sun": 3})
data["time"] = data["time"].map({"Lunch": 0, "Dinner": 1})
data.head(5)
```

```python
x = np.array(data[["total_bill", "sex", "smoker", "day", "time", "size"]])
y = np.array(data["tip"])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

### Features = [[total_bill, "sex", "smoker", "day", "time", "size"]]

### 04. Model Performance Evaluation

**Model Metrics:** Evaluate the Linear Regression model using various performance metrics including R² Score, Mean Squared Error (MSE), Mean Absolute Error (MAE), and Root Mean Squared Error (RMSE).

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import math

# Generate predictions on training and test data
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)

# Calculate evaluation metrics
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)
mse_train = mean_squared_error(y_train, y_train_pred)
mse_test = mean_squared_error(y_test, y_test_pred)
mae_train = mean_absolute_error(y_train, y_train_pred)
mae_test = mean_absolute_error(y_test, y_test_pred)
rmse_train = math.sqrt(mse_train)
rmse_test = math.sqrt(mse_test)

# Display model evaluation metrics
print("=" * 60)
print("MODEL PERFORMANCE METRICS")
print("=" * 60)
print(f"\n{'Metric':<20} {'Training':<20} {'Testing':<20}")
print("-" * 60)
print(f"{'R² Score':<20} {r2_train:<20.4f} {r2_test:<20.4f}")
print(f"{'Mean Squared Error':<20} {mse_train:<20.4f} {mse_test:<20.4f}")
print(f"{'Root Mean Sq. Error':<20} {rmse_train:<20.4f} {rmse_test:<20.4f}")
print(f"{'Mean Absolute Error':<20} {mae_train:<20.4f} {mae_test:<20.4f}")
print("=" * 60)
```

```python
# Create comparison DataFrame for test set predictions
comparison_df = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_test_pred,
    'Residuals': y_test - y_test_pred,
    'Absolute Error': np.abs(y_test - y_test_pred)
})

print("\nPrediction Comparison (First 10 Test Samples):")
print(comparison_df.head(10).to_string())
print(f"\nTotal Test Samples: {len(comparison_df)}")
```

- total_bill: Total bill amount in dollars

- smoker: 0 = No, 1 = Yes
- day: 0 = Thursday, 1 = Friday, 2 = Saturday, 3 = Sunday
- time: 0 = Lunch, 1 = Dinner
- size: Number of people in the party

### 07. Model Summary & Interpretation

**Model Type:** Linear Regression

**Key Findings:**
- **R² Score (Test):** The model explains approximately {:.1%} of the variance in tip amounts
- **RMSE (Test):** Average prediction error is approximately ${:.2f}
- **MAE (Test):** Mean absolute error shows typical prediction deviation

**Model Coefficients:**
- The model weights each feature based on its contribution to predicting tips
- Positive coefficients increase predicted tips
- Negative coefficients decrease predicted tips

**Performance Analysis:**
- If R² is close to 1.0: The model explains most variance (good fit)
- If R² is close to 0.0: The model has poor predictive power
- RMSE and MAE indicate average magnitude of prediction errors in dollars

```python
# Display Model Coefficients
print("\n" + "=" * 60)
print("MODEL COEFFICIENTS & FEATURE IMPORTANCE")
print("=" * 60)

feature_names = ["total_bill", "sex", "smoker", "day", "time", "size"]
coefficients = model.coef_
intercept = model.intercept_

print(f"\nIntercept (Base Prediction): ${intercept:.4f}\n")
print(f"{'Feature':<15} {'Coefficient':<15} {'Interpretation':<30}")
print("-" * 60)

for feature, coef in zip(feature_names, coefficients):
    direction = "increases" if coef > 0 else "decreases"
    impact = abs(coef)
    print(f"{feature:<15} {coef:<15.6f} {direction} tips by ${impact:.4f}")

print("=" * 60)
print(f"\nPrediction Formula:")
print(f"Tip = {intercept:.4f} + ", end="")
print(" + ".join([f"{coef:.4f}*{feat}" for feat, coef in zip(feature_names, coefficients)]))
```
